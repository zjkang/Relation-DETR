```python
import matplotlib.pyplot as plt
import numpy as np

# 数据
heads = [0, 2, 4, 6,8]
ap = [51.1, 51.4, 51.8, 52.0,52.3]

# 创建图形，设置dpi可以控制输出图像的质量
plt.figure(figsize=(10, 6), dpi=300)

# 设置y轴范围使差异更明显
plt.ylim(50, 53)

# 绘制柱状图
bars = plt.bar(heads, ap, 
               width=0.8,  # 调整柱子宽度
               color='#7CB9E8',  # 淡青色
               # 或者使用渐变色：
               # color=['#2878B5', '#4B97C5', '#6FB1D4', '#8FCAE2']
               )
# bars = plt.bar(heads, ap，color='#72C6C6')

# 美化设置
plt.xlabel('Number of Relation Heads', fontsize=12)
plt.ylabel('AP', fontsize=12)
plt.title('AP vs Number of Relation Heads', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.3)

# 在柱子上添加数值标签
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.1f}',
             ha='center', va='bottom')

# 设置x轴刻度
plt.xticks(heads)

plt.tight_layout()

# 保存图像
# dpi控制分辨率，bbox_inches='tight'确保完整保存，transparent=False设置不透明背景
# plt.savefig('relation_heads_ap.png', dpi=300, bbox_inches='tight', transparent=False)
# 也可以保存为PDF以获得矢量图
plt.savefig('relation_heads_ap.pdf', bbox_inches='tight')

plt.show()
```

```python
import cv2
import torch

from matplotlib import pyplot as plt
from util.utils import load_state_dict, load_checkpoint
from util.visualize import plot_bounding_boxes_on_image
```

## 2. Load a model and class information

We encode class information into the checkpoint, use our defined `load_state_dict` to handle it.

```python
from configs.relation_detr.relation_detr_resnet50_800_1333 import model
weight = load_checkpoint("https://github.com/xiuqhou/Relation-DETR/releases/download/v1.0.0/relation_detr_resnet50_800_1333_coco_1x.pth")
load_state_dict(model, weight)
model = model.eval()
```

```python
image = cv2.imread("data/coco/val2017/000000000139.jpg")
torch_image = torch.tensor(image.transpose(2, 0, 1))
```

## 3. Inference on the image

```python
with torch.no_grad():
    predictions = model([torch_image])[0]
# The following code also works:
# predictions = model(torch_image.unsqueeze(0))[0]
```

## 4. Visualize results

```python
image_for_show = plot_bounding_boxes_on_image(
    image=cv2.cvtColor(image, cv2.COLOR_BGR2RGB),
    boxes=predictions["boxes"],
    labels=predictions["labels"],
    scores=predictions["scores"],
    classes=model.CLASSES,  # class information
    show_conf=0.5,
    font_scale=1.0,
    box_thick=2,
    text_alpha=0.75,
)
```
